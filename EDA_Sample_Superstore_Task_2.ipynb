### Objective
Explore the Superstore dataset to identify important patterns, relationships, outliers, and trends using Python, Pandas, Matplotlib, and Seaborn.



## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")

sns.set_theme(style="whitegrid")

print("Libraries imported successfully.")
```

## 2. Load the Dataset



```python
FILE_NAME = "Sample - Superstore.csv"

df = pd.read_csv(FILE_NAME, encoding="latin1")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

## 3. Initial Data Inspection

```python
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
display(df.dtypes)

print("\nMissing values:")
display(df.isnull().sum().sort_values(ascending=False))

print("\nDuplicate rows:", df.duplicated().sum())
```

### Initial Observation

The initial inspection helps identify the size of the dataset, data types, missing values, and possible duplicate records before performing statistical or visual analysis.


## 4. Data Cleaning and Preparation

```python

df = df.drop_duplicates().copy()

df["Order Date"] = pd.to_datetime(df["Order Date"], errors="coerce")

# Create useful time columns
df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
df["Year-Month"] = df["Order Date"].dt.to_period("M").astype(str)


numeric_cols = ["Sales", "Quantity", "Discount", "Profit"]

print("Missing values after basic cleaning:")
display(df[numeric_cols].isnull().sum())

print("\nCleaned shape:", df.shape)
```

## 5. Summary Statistics

`.describe()` provides the central tendency, spread, minimum, maximum, and quartiles of the numerical variables.


```python
summary_stats = df[["Sales", "Quantity", "Discount", "Profit"]].describe()
display(summary_stats)
```

### Observation

The summary statistics show the typical scale and variation of Sales, Quantity, Discount, and Profit. The difference between the mean and median can also indicate skewness in the distributions.


## 6. Correlation Analysis

### Question
**How strongly are Sales, Quantity, Discount, and Profit related to each other?**


```python
corr = df[["Sales", "Quantity", "Discount", "Profit"]].corr()

plt.figure(figsize=(9, 6))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Correlation Heatmap of Numerical Variables")
plt.tight_layout()
plt.show()
```

### Observation

The correlation heatmap helps identify positive and negative linear relationships. In particular, Sales and Profit can be examined to determine whether higher sales generally correspond to higher profit, while Discount can be checked for its relationship with Profit.


## 7. Visualization 1 — Distribution of Sales

### Question
**How are individual order sales distributed?**


```python
plt.figure(figsize=(10, 6))
sns.histplot(df["Sales"], bins=40, kde=True)
plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Number of Orders")
plt.tight_layout()
plt.show()
```

### Observation

The Sales distribution is typically right-skewed, meaning that most orders have relatively smaller sales values while a smaller number of orders have much higher sales. This is useful when understanding why the mean can be higher than the median.


## 8. Visualization 2 — Sales and Profit by Category

### Question
**Which product categories generate the most Sales and Profit?**


```python
category_summary = (
    df.groupby("Category")[["Sales", "Profit"]]
      .sum()
      .sort_values("Sales", ascending=False)
)

display(category_summary)

category_summary.plot(kind="bar", figsize=(10, 6))
plt.title("Total Sales and Profit by Category")
plt.xlabel("Category")
plt.ylabel("Amount")
plt.xticks(rotation=0)
plt.legend(title="Metric")
plt.tight_layout()
plt.show()
```

### Observation

The category comparison shows how revenue generation differs from profitability. A category with high Sales does not automatically have the highest Profit, so both measures should be considered together.


## 9. Visualization 3 — Sales vs Profit

### Question
**Does higher Sales generally correspond to higher Profit?**


```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x="Sales",
    y="Profit",
    hue="Category",
    alpha=0.6
)
plt.axhline(0, linestyle="--", linewidth=1)
plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.tight_layout()
plt.show()
```

### Observation

The scatter plot reveals the relationship between order-level Sales and Profit. Points below zero represent loss-making orders. This visualization is especially useful for identifying cases where relatively high Sales do not result in positive Profit.


## 10. Visualization 4 — Outliers in Sales, Discount, and Profit

### Question
**Are there unusual observations that may require further investigation?**


```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

sns.boxplot(y=df["Sales"], ax=axes[0])
axes[0].set_title("Sales Outliers")
axes[0].set_ylabel("Sales")

sns.boxplot(y=df["Discount"], ax=axes[1])
axes[1].set_title("Discount Outliers")
axes[1].set_ylabel("Discount")

sns.boxplot(y=df["Profit"], ax=axes[2])
axes[2].set_title("Profit Outliers")
axes[2].set_ylabel("Profit")

plt.tight_layout()
plt.show()
```

### Observation

The boxplots highlight unusually large or small observations. These points should not automatically be deleted: an outlier can represent a genuine high-value order, unusually high discount, or substantial loss and may contain useful business information.


## 11. Visualization 5 — Monthly Sales Trend

### Question
**How does Sales change over time?**


```python
monthly_sales = (
    df.groupby("Year-Month")["Sales"]
      .sum()
      .reset_index()
)

plt.figure(figsize=(14, 6))
sns.lineplot(data=monthly_sales, x="Year-Month", y="Sales", marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Year-Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=60)
plt.tight_layout()
plt.show()
```

### Observation

The monthly trend makes it possible to identify periods of increasing or decreasing sales and recurring peaks. Such patterns can help businesses investigate seasonality and plan inventory or marketing activities.


## 12. Visualization 6 — Regional Performance

### Question
**Which regions contribute the most Sales and Profit?**


```python
region_summary = (
    df.groupby("Region")[["Sales", "Profit"]]
      .sum()
      .sort_values("Sales", ascending=False)
)

display(region_summary)

region_summary.plot(kind="bar", figsize=(10, 6))
plt.title("Sales and Profit by Region")
plt.xlabel("Region")
plt.ylabel("Amount")
plt.xticks(rotation=0)
plt.legend(title="Metric")
plt.tight_layout()
plt.show()
```

### Observation

Regional aggregation highlights differences in business performance across geographic areas. Comparing Sales and Profit together helps distinguish regions that generate revenue from regions that generate profitable revenue.


## 13. Additional Analysis — Discount vs Profit

### Question
**Is there a visible relationship between discount levels and profitability?**


```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x="Discount",
    y="Profit",
    hue="Category",
    alpha=0.6
)
plt.axhline(0, linestyle="--", linewidth=1)
plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.tight_layout()
plt.show()
```

### Observation

The relationship between Discount and Profit can reveal whether heavily discounted orders are associated with lower profitability. The pattern should be interpreted as an association rather than proof that discount alone causes a change in profit.
